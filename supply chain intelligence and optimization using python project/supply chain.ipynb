```python
# Analyze profitability of each product category

# Calculate total purchase cost
purchase_order_items["total_purchase"] = (
    purchase_order_items["quantity_ordered"] *
    purchase_order_items["unit_cost"]
)

# Add category information
purchase = purchase_order_items.merge(
    products[["product_id", "category"]],
    on="product_id"
)

# Calculate category-wise purchase cost
category_cost = purchase.groupby("category")["total_purchase"].sum()


# Calculate total sales revenue
sales_order_items["total_revenue"] = (
    sales_order_items["quantity"] *
    sales_order_items["unit_selling_price"]
)

# Add category information
sales = sales_order_items.merge(
    products[["product_id", "category"]],
    on="product_id"
)

# Calculate category-wise revenue
category_revenue = sales.groupby("category")["total_revenue"].sum()


# Combine cost and revenue
category_analysis = pd.concat(
    [category_cost, category_revenue], axis=1
).fillna(0)

category_analysis.columns = ["total_purchase", "total_revenue"]

# Calculate profit
category_analysis["total_profit"] = (
    category_analysis["total_revenue"] -
    category_analysis["total_purchase"]
)

# Sort by profit
category_analysis = category_analysis.sort_values(
    by="total_profit", ascending=False
)

# Calculate average profit
average_profit = category_analysis["total_profit"].mean()

# Business optimization
category_analysis["recommendation"] = category_analysis["total_profit"].apply(
    lambda x: "High Priority - Focus on this profitable category and maintain sufficient stock."
    if x >= average_profit
    else "Medium Priority - Reduce purchasing costs and improve sales through promotions."
)

pd.set_option("display.float_format", "{:,.0f}".format)

print(category_analysis.to_string())
```

```python

plt.figure(figsize=(10, 5))

plt.bar(
    category_analysis.index,
    category_analysis["total_profit"]
)

plt.title("Category-wise Profit")
plt.xlabel("Product Category")
plt.ylabel("Profit (Crores)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Analyze the yearly sales revenue trend

# Calculate total revenue
sales_order_items["total_revenue"] = (
    sales_order_items["quantity"] *
    sales_order_items["unit_selling_price"]
)

# Merge sales orders
yearly_sales = sales_order_items.merge(
    sales_orders[["sales_order_id", "order_date"]],
    on="sales_order_id"
)

# Extract year
yearly_sales["year"] = pd.to_datetime(
    yearly_sales["order_date"], dayfirst=True
).dt.year

# Calculate yearly revenue
yearly_revenue = yearly_sales.groupby(
    "year"
)["total_revenue"].sum().reset_index()

yearly_revenue = yearly_revenue.sort_values("year")


print(yearly_revenue.to_string(index=False))
```
